# 3️⃣  Data Cleaning

```python
# ============ 1) Copy ============
# نشتغل على نسخة علشان الأصل يفضل سليم
# Work on copies so the originals stay untouched
tr = train.copy()
st = store.copy()
```

```python
# ============ 2) Closed days ============
# الأيام المقفولة مبيعاتها كلها صفر، ومش بتعلّم الموديل حاجة
# Closed days always have zero sales, so they teach the model nothing
print("closed days:", (tr["Open"] == 0).sum())   # المتوقع / expected: 172817
tr = tr[tr["Open"] == 1]
```

```python
# ============ 3) Open but zero sales ============
# أيام مفتوحة ومبيعاتها صفر (54 صف) شاذة، فبنشيلها
# Open days with zero sales (54 rows) are anomalies, so we drop them
print("open & zero sales:", (tr["Sales"] == 0).sum())   # المتوقع / expected: 54
tr = tr[tr["Sales"] > 0]
```

```python
# ============ 4) Drop Open ============
# بعد الفلترة العمود ده كله 1، فمالوش لازمة
# After filtering, this column is all 1s, so it is useless
tr = tr.drop(columns=["Open"])
```

```python
# ============ 5) Sort ============
# ترتيب بالفرع ثم التاريخ، ده مهم للـ Lag بعدين
# Sort by store then date, this matters for lag features later
tr = tr.sort_values(["Store", "Date"]).reset_index(drop=True)
```

```python
# ============ 6) Store: CompetitionDistance ============
# 3 فروع بس ناقصهم المسافة، فبنملاها بالـ median (مش بتتأثر بالقيم الكبيرة)
# Only 3 stores miss the distance, so we fill with the median (robust to outliers)
st["CompetitionDistance"] = st["CompetitionDistance"].fillna(st["CompetitionDistance"].median())
```

```python
# ============ 7) Store: CompetitionOpenSince ============
# الناقص هنا معناه "مفيش معلومة"، فبنملاه بصفر ونحوله لميزة في الـ Feature Engineering
# Missing here means "no info", so we fill with 0 and handle it in Feature Engineering
for c in ["CompetitionOpenSinceMonth", "CompetitionOpenSinceYear"]:
    st[c] = st[c].fillna(0).astype(int)
```

```python
# ============ 8) Store: Promo2 columns ============
# الناقص هنا مش خطأ، ده الفروع اللي Promo2 = 0 أصلًا
# Missing here is not an error, these are stores with Promo2 = 0
for c in ["Promo2SinceWeek", "Promo2SinceYear"]:
    st[c] = st[c].fillna(0).astype(int)
st["PromoInterval"] = st["PromoInterval"].fillna("None")
```

```python
# ============ 9) Check ============
# نتأكد إن مفيش Missing باقي
# Make sure no missing values remain
print(tr.isna().sum().sum(), st.isna().sum().sum())   # المتوقع / expected: 0 0
print("train after cleaning:", tr.shape)
```

```python
# ============ 10) Save ============
# نحفظ الملفين النضاف علشان نرفعهم PostgreSQL في الخطوة الجاية
# Save the clean files to load into PostgreSQL in the next step
tr.to_csv("train_clean.csv", index=False)
st.to_csv("store_clean.csv", index=False)
```

```python
# ============ 1) Connection ============
# بيانات الاتصال بقاعدة البيانات (غيّر الباسورد)
# Database connection details (change the password)
username = 'postgres'
password = '1999'
host = 'localhost'
port = '5432'
database = 'rossmann'
engine = create_engine(f'postgresql://{username}:{password}@{host}:{port}/{database}')
```

```python
# ============ 2) Load clean files ============
# تحميل الملفين النضاف من خطوة الـ Cleaning
# Load the clean files from the Cleaning step
train_clean = pd.read_csv("train_clean.csv", parse_dates=["Date"], dtype={"StateHoliday": "string"})
store_clean = pd.read_csv("store_clean.csv")

# أسماء الأعمدة بحروف صغيرة علشان Postgres ما يطلبش Quotes
# Lowercase column names so Postgres does not require quotes
train_clean.columns = train_clean.columns.str.lower()
store_clean.columns = store_clean.columns.str.lower()
```

```python
# ============ 3) Upload tables ============
# قاموس فيه الجداول اللي هنرفعها
# Dictionary of the tables we will upload
datasets_clean = {
    'store': store_clean,
    'train': train_clean
}

# رفع كل جدول، والـ chunksize علشان train كبير (844 ألف صف) فيترفع على دفعات
# Upload each table; chunksize because train is large (844K rows) so it uploads in batches
for table_name, df in datasets_clean.items():
    df.to_sql(table_name, engine, if_exists='replace', index=False, chunksize=50000)
    print(f"✅ Upload Done {table_name} — {df.shape[0]} Row")
```

```python
# ============ 4) Index ============
# Primary Key على الفرع في store و Index على train علشان الـ Join يبقى أسرع
# Primary key on store and an index on train to make the join faster
with engine.begin() as conn:
    conn.execute(text("ALTER TABLE store ADD PRIMARY KEY (store);"))
    conn.execute(text("CREATE INDEX IF NOT EXISTS idx_train_store ON train(store);"))
```

```python
# ============ 5) Join ============
# ربط الجدولين على رقم الفرع
# Join the two tables on the store ID
query = """
SELECT
    t.store,
    t.date,
    t.dayofweek,
    t.sales,
    t.customers,
    t.promo,
    t.stateholiday,
    t.schoolholiday,
    s.storetype,
    s.assortment,
    s.competitiondistance,
    s.competitionopensincemonth,
    s.competitionopensinceyear,
    s.promo2,
    s.promo2sinceweek,
    s.promo2sinceyear,
    s.promointerval
FROM train t
JOIN store s ON t.store = s.store
ORDER BY t.store, t.date
"""

main_df = pd.read_sql(query, engine, parse_dates=["date"])

# المفروض نفس عدد صفوف train بالظبط (844338) و17 عمود
# Should be exactly the train row count (844338) and 17 columns
print("Shape:", main_df.shape)
print(main_df.head())
```

```python
# ============ 6) Save ============
# نحفظ النتيجة للخطوات الجاية
# Save the result for the next steps
main_df.to_csv("rossmann_merged.csv", index=False)
```

# 6️⃣ Feature Engineering Part A   — Key Insights¶

```python
# ============ 1) Load ============
# تحميل الداتا المدموجة (مرتبة بالفرع ثم التاريخ)
# Load the merged data (sorted by store then date)
df = pd.read_csv("rossmann_merged.csv", parse_dates=["date"], dtype={"stateholiday": "string"})

# إصلاح: pandas بيقرا "None" كـ NaN، فبنرجّعها نص
# Fix: pandas reads "None" as NaN, so we turn it back into a string
df["promointerval"] = df["promointerval"].fillna("None")

df = df.sort_values(["store", "date"]).reset_index(drop=True)
```

```python
# ============ 2) Date features ============
# استخراج السنة والشهر واليوم ورقم الأسبوع والربع من التاريخ
# Extract year, month, day, week of year and quarter from the date
# (dayofweek موجود أصلًا في الداتا / dayofweek already exists in the data)
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["weekofyear"] = df["date"].dt.isocalendar().week.astype(int)
df["quarter"] = df["date"].dt.quarter
```

```python
# ============ 3) State holiday flag ============
# علامة بسيطة: في إجازة رسمية ولا لا (النوع التفصيلي هنتعامل معاه في الـ Encoding)
# Simple flag: is it a state holiday or not (the detailed type is handled in encoding)
df["is_state_holiday"] = (df["stateholiday"] != "0").astype(int)
```

```python
# ============ 4) Competition: months since open ============
# عدد الأشهر من وقت فتح المنافس لحد تاريخ اليوم
# Number of months between the competitor opening and the current date
df["competition_open_months"] = (
    12 * (df["year"] - df["competitionopensinceyear"])
    + (df["month"] - df["competitionopensincemonth"])
)

# السنة 0 معناها مفيش معلومة، فنحط 0 ونضيف علامة توضح ده
# Year 0 means "no info", so we set 0 and add a flag to mark it
df.loc[df["competitionopensinceyear"] == 0, "competition_open_months"] = 0
df["has_competition_info"] = (df["competitionopensinceyear"] > 0).astype(int)

# لو المنافس هيفتح في المستقبل (قيمة سالبة) نخليها 0
# If the competitor opens in the future (negative value) we set it to 0
df["competition_open_months"] = df["competition_open_months"].clip(lower=0)
```

```python
# ============ 5) Promo2 active ============
# هل Promo2 شغالة فعلًا في اليوم ده؟ لازم 3 شروط: الفرع مشترك، بدأت فعلًا، والشهر ضمن الـ PromoInterval
# Is Promo2 really running on this day? Needs 3 conditions: store enrolled, already started, month in PromoInterval
promo2_start = (
    pd.to_datetime(df["promo2sinceyear"].astype(str) + "-01-01", format="%Y-%m-%d", errors="coerce")
    + pd.to_timedelta((df["promo2sinceweek"] - 1).clip(lower=0) * 7, unit="D")
)

# تنبيه: في الداتا سبتمبر مكتوبة "Sept" مش "Sep"
# Note: in the data September is written "Sept", not "Sep"
month_str = df["date"].dt.strftime("%b").replace({"Sep": "Sept"})
month_in_interval = np.array([m in iv.split(",") for m, iv in zip(month_str, df["promointerval"])])

df["promo2_active"] = (
    (df["promo2"] == 1) & (df["date"] >= promo2_start) & month_in_interval
).astype(int)
```

```python
# ============ 6) Drop Customers ============
# حذف Customers علشان الـ Data Leakage (مش معروف وقت التوقع)
# Drop Customers because of data leakage (unknown at prediction time)
df = df.drop(columns=["customers"])
```

```python
# ============ 7) Check ============
# نتأكد من الشكل ومن عدم وجود Missing
# Check the shape and make sure there are no missing values
print("Shape:", df.shape)                       # المتوقع / expected: (844338, 25)
print("Missing:", df.isna().sum().sum())        # المتوقع / expected: 0
print(df[["year", "month", "weekofyear", "competition_open_months", "promo2_active"]].head())
```

 # Feature Engineering (Part B: Lag & Rolling) — Key Insights

```python
# ============ 1) Load ============
# تحميل نتيجة الجزء A
# Load the result of part A
df = pd.read_csv("rossmann_fe_part1.csv", parse_dates=["date"], dtype={"stateholiday": "string"})
df["promointerval"] = df["promointerval"].fillna("None")   # pandas بيقرا "None" كـ NaN / pandas reads "None" as NaN
df = df.sort_values(["store", "date"]).reset_index(drop=True)
```

```python
# ============ 2) Wide table ============
# جدول عرضه الفروع وطوله التواريخ، وبنكمّل كل الأيام (الأيام المقفولة بتبقى NaN)
# Table with stores as columns and dates as rows; we fill in all days (closed days become NaN)
wide = df.pivot(index="date", columns="store", values="sales")
wide = wide.reindex(pd.date_range(wide.index.min(), wide.index.max(), freq="D"))
wide
```

```python
# ============ 3) Features ============
# past: المبيعات متأخرة 7 أيام، علشان نبص على الماضي بس (مفيش Leakage)
# past: sales shifted by 7 days so we only look at the past (no leakage)
past = wide.shift(7)

feats = {
    # مبيعات نفس اليوم من أسبوع ومن أسبوعين
    # Sales on the same day one week and two weeks ago
    "lag_7": wide.shift(7),
    "lag_14": wide.shift(14),

    # متوسط 7 و30 يوم (الـ NaN بيتتجاهل، و min_periods=1 يعني يكفي يوم واحد)
    # 7-day and 30-day averages (NaN is ignored; min_periods=1 means one day is enough)
    "rolling_mean_7": past.rolling(7, min_periods=1).mean(),
    "rolling_mean_30": past.rolling(30, min_periods=1).mean(),
}
```

```python
# ============ 4) Merge back ============
# نحوّل كل Feature لشكل طويل ونربطه بالداتا على (التاريخ + الفرع)
# Convert each feature to long format and merge it on (date + store)
for name, f in feats.items():
    long = f.rename_axis("date").reset_index().melt(id_vars="date", var_name="store", value_name=name)
    df = df.merge(long, on=["date", "store"], how="left")
```

```python
# ============ 5) Check ============
# المفروض الصفوف ما تتغيرش، والـ NaN موجود بس في الأيام الأولى أو لو اليوم المرجعي كان مقفول
# Rows should not change; NaN appears only in the first days or when the reference day was closed
print("Shape:", df.shape)            # المتوقع / expected: (844338, 29)
print(df[["lag_7", "lag_14", "rolling_mean_7", "rolling_mean_30"]].isna().sum())
print(df[["lag_7", "rolling_mean_7"]].corrwith(df["sales"]).round(3))
```

```python
# ============ 6) Save ============
# حفظ الداتا النهائية لخطوة الـ ML
# Save the final data for the ML step
df.to_csv("rossmann_fe_final.csv", index=False)
```

# 7️⃣ StateHoliday Analysis (Alternative to NLP)

```python
import pandas as pd
import numpy as np

# ============ 1) Load raw data ============
# بنقرأ الداتا الأصلية (قبل ما نشيل الأيام المقفولة) علشان نشوف كام فرع بيقفل
# Load the raw data (before removing closed days) to see how many stores close
raw = pd.read_csv("train.csv", low_memory=False, parse_dates=["Date"], dtype={"StateHoliday": "string"})
raw.columns = raw.columns.str.lower()
```

```python
# ============ 2) Holiday types: how many stores close? ============
# لكل نوع إجازة: عدد الصفوف، نسبة الفروع المقفولة، ومتوسط مبيعات الفروع اللي فتحت
# For each holiday type: row count, share of closed stores, average sales of stores that opened
summary = raw.groupby("stateholiday").agg(
    rows=("sales", "size"),
    closed_share=("open", lambda s: 1 - s.mean()),
    mean_sales_open=("sales", lambda s: s[s > 0].mean()))
print(summary.round(3))
```

```python
# ============ 3) Store-adjusted effect ============
# بنقارن كل فرع بنفسه: مبيعات اليوم ÷ متوسط نفس الفرع في نفس اليوم من الأسبوع ونفس حالة العرض (في الأيام العادية)
# We compare each store with itself: sales ÷ the same store's average on the same weekday and promo status (normal days)
op = raw[(raw["open"] == 1) & (raw["sales"] > 0)].copy()
baseline = (op[op["stateholiday"] == "0"]
            .groupby(["store", "dayofweek", "promo"])["sales"].mean().rename("baseline"))
op = op.join(baseline, on=["store", "dayofweek", "promo"])
op["ratio"] = op["sales"] / op["baseline"]
print(op.groupby("stateholiday")["ratio"].agg(["mean", "count"]).round(3))
```

```python
# ============ 4) Day before / after a holiday ============
# تأثير اليوم اللي قبل الإجازة واللي بعدها (الأيام العادية بس)
# Effect of the day before and the day after a holiday (normal days only)
all_holidays = pd.Series(raw.loc[raw["stateholiday"] != "0", "date"].unique())
is_normal = op["stateholiday"] == "0"
before = (op["date"] + pd.Timedelta(days=1)).isin(all_holidays)
after = (op["date"] - pd.Timedelta(days=1)).isin(all_holidays)
print("day before holiday:", round(op.loc[is_normal & before, "ratio"].mean(), 3))
print("day after holiday :", round(op.loc[is_normal & after, "ratio"].mean(), 3))
```

```python
# ============ 5) Holiday calendar features ============
# تقويم الإجازات معروف مقدمًا، فمفيش Leakage. بنضيف test.csv علشان نعرف إجازات أغسطس وسبتمبر 2015
# The holiday calendar is known in advance, so no leakage. We add test.csv to know the Aug-Sep 2015 holidays
df = pd.read_csv("rossmann_fe_final.csv", parse_dates=["date"], dtype={"stateholiday": "string"})
test_cal = pd.read_csv("test.csv", parse_dates=["Date"], dtype={"StateHoliday": "string"})
holiday_dates = np.array(sorted(set(all_holidays) |
                                set(test_cal.loc[test_cal["StateHoliday"].fillna("0") != "0", "Date"])),
                         dtype="datetime64[ns]")

# عدد الأيام لحد الإجازة الجاية، وعدد الأيام من آخر إجازة
# Days until the next holiday, and days since the last holiday
d = df["date"].values.astype("datetime64[ns]")
i = np.searchsorted(holiday_dates, d)
to_next = np.where(i < len(holiday_dates),
                   (holiday_dates[np.minimum(i, len(holiday_dates) - 1)] - d) / np.timedelta64(1, "D"), 99)
j = np.searchsorted(holiday_dates, d, side="right") - 1
since_last = np.where(j >= 0, (d - holiday_dates[np.maximum(j, 0)]) / np.timedelta64(1, "D"), 99)

# بنحدد السقف بـ 14 يوم (بعد كده الإجازة مالهاش تأثير)
# Cap at 14 days (beyond that the holiday has no effect)
df["days_to_next_holiday"] = np.minimum(to_next, 14)
df["days_since_last_holiday"] = np.minimum(since_last, 14)
df["day_before_holiday"] = (to_next == 1).astype(int)
df["day_after_holiday"] = (since_last == 1).astype(int)
print(df.shape)   # المتوقع / expected: (844338, 33)
```

```python
# ============ 6) Save ============
# حفظ النسخة الجديدة للـ ML
# Save the new version for the ML step
df.to_csv("rossmann_fe_final_v2.csv", index=False)
```

# 8️⃣ Machine Learning 🤖

```python
import pickle
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit
from sklearn.metrics import mean_squared_error, mean_absolute_error
from xgboost import XGBRegressor
```

```python
# ============ 1) Load ============
# لازم نقرا النسخة الجديدة v2 (فيها 33 عمود)
# We must read the new v2 file (it has 33 columns)
df = pd.read_csv("rossmann_fe_final_v2.csv", parse_dates=["date"], dtype={"stateholiday": "string"})
df["promointerval"] = df["promointerval"].fillna("None")
df = df.sort_values(["date", "store"]).reset_index(drop=True)
print("df shape:", df.shape)    # المتوقع / expected: (844338, 33)

```

```python
# ============ 2) Time-based split ============
# نعيد التقسيم من الـ df الجديد
# Redo the split from the new df
val_start = pd.Timestamp("2015-05-08")
test_start = pd.Timestamp("2015-06-19")
train = df[df["date"] < val_start]
val = df[(df["date"] >= val_start) & (df["date"] < test_start)]
test = df[df["date"] >= test_start]
print(train.shape, val.shape, test.shape)   # المتوقع / expected: (765470, 33) (37472, 33) (41396, 33)
```

```python
# ============ 3) Features ============
# القائمة لازم تحتوي على الـ 4 الجديدة في آخرها
# The list must include the 4 new ones at the end
numerical_features = ["year", "month", "day", "weekofyear", "quarter", "dayofweek",
                      "promo", "schoolholiday", "is_state_holiday",
                      "competitiondistance", "competition_open_months", "has_competition_info",
                      "promo2", "promo2_active",
                      "lag_7", "lag_14", "rolling_mean_7", "rolling_mean_30",
                      "days_to_next_holiday", "days_since_last_holiday",
                      "day_before_holiday", "day_after_holiday"]
categorical_features = ["stateholiday", "storetype", "assortment", "promointerval"]

x_train, y_train = train[numerical_features + categorical_features], train["sales"]
x_val, y_val = val[numerical_features + categorical_features], val["sales"]
x_test, y_test = test[numerical_features + categorical_features], test["sales"]

# ============ Check ============
# نتأكد إن الأعمدة الجديدة دخلت فعلًا قبل التدريب
# Make sure the new columns are really inside before training
print(x_train.shape[1])   # المتوقع / expected: 22 + 4 = 26

```

```python
# ============ 4) Preprocessing ============
# نفس أسلوبك: Pipeline للأرقام وPipeline للنصوص جوه ColumnTransformer
# Same as your style: a pipeline for numbers and one for text inside a ColumnTransformer
numerical_pipe = Pipeline(steps=[
    ("imputing", SimpleImputer(strategy="median")),
    ("scaling", StandardScaler())
])
categorical_pipe = Pipeline(steps=[
    ("imputing", SimpleImputer(strategy="most_frequent")),
    ("encoding", OneHotEncoder(drop="first", handle_unknown="ignore"))
])
preprocessing = ColumnTransformer(transformers=[
    ("numerical_processing", numerical_pipe, numerical_features),
    ("categorical_processing", categorical_pipe, categorical_features)
])
```

```python
# ============ 5) Helpers ============
# دالة بتبني Pipeline كامل، والموديل بيتدرب على log(sales) ويرجّع المبيعات الحقيقية لوحده
# Builds a full pipeline; the model trains on log(sales) and returns real sales by itself
def make_pipeline(model):
    return TransformedTargetRegressor(
        regressor=Pipeline(steps=[("preprocess", preprocessing), ("model", model)]),
        func=np.log1p, inverse_func=np.expm1
    )

# RMSPE هو المقياس الرسمي للمسابقة (خطأ نسبي)
# RMSPE is the official competition metric (relative error)
def rmspe(y_true, y_pred):
    return np.sqrt(np.mean(((y_true - y_pred) / y_true) ** 2))

# نعرّفها مرة واحدة بس (مش في كل خلية)
# Defined once only (not in every cell)
def evaluate(name, pipe, x, y, tag):
    pred = pipe.predict(x)
    print(f"{name:14s} {tag:5s} RMSE={np.sqrt(mean_squared_error(y, pred)):7.1f}  "
          f"MAE={mean_absolute_error(y, pred):7.1f}  RMSPE={rmspe(y.values, pred):.4f}")
    return rmspe(y.values, pred)

val_scores = {}

# TimeSeriesSplit بدل cv=5: كل Fold بيتدرب على الماضي ويتجرب على اللي بعده
# TimeSeriesSplit instead of cv=5: each fold trains on the past and tests on what follows
tscv = TimeSeriesSplit(n_splits=3)
```

```python
# ============ 6) Linear Regression (Baseline) ============
# الموديل الأساسي للمقارنة
# The baseline model for comparison
LR_pip = make_pipeline(LinearRegression()).fit(x_train, y_train)
evaluate("LinearReg", LR_pip, x_train, y_train, "Train")
val_scores["LinearReg"] = evaluate("LinearReg", LR_pip, x_val, y_val, "Val")
```

```python
# ============ 7) Decision Tree (Tuned) ============
# ضبط العمق بالـ GridSearchCV الزمني
# Tune the depth with the time-based GridSearchCV
dt_grid = GridSearchCV(make_pipeline(DecisionTreeRegressor(random_state=42)),
                       param_grid={"regressor__model__max_depth": [8, 12, 16]},
                       cv=tscv, scoring="neg_root_mean_squared_error")
dt_grid.fit(x_train, y_train)
print("DT best:", dt_grid.best_params_)
DT_pip = dt_grid.best_estimator_
evaluate("DecisionTree", DT_pip, x_train, y_train, "Train")
val_scores["DecisionTree"] = evaluate("DecisionTree", DT_pip, x_val, y_val, "Val")
```

```python
# ============ 8) Random Forest ============
# من غير Grid علشان الوقت (حوالي دقيقتين)
# No grid because of time (about 2 minutes)
RF_pip = make_pipeline(RandomForestRegressor(
    n_estimators=100, max_depth=16, min_samples_leaf=5,
    max_features=0.5, n_jobs=-1, random_state=42)).fit(x_train, y_train)
evaluate("RandomForest", RF_pip, x_train, y_train, "Train")
val_scores["RandomForest"] = evaluate("RandomForest", RF_pip, x_val, y_val, "Val")
```

```python
# ============ 9) XGBoost (Tuned) ============
# ضبط العمق وعدد الأشجار
# Tune depth and number of trees
xgb_grid = GridSearchCV(
    make_pipeline(XGBRegressor(tree_method="hist", subsample=0.8, colsample_bytree=0.8,
                               learning_rate=0.1, n_jobs=-1, random_state=42)),
    param_grid={"regressor__model__max_depth": [6, 8],
                "regressor__model__n_estimators": [200, 400]},
    cv=tscv, scoring="neg_root_mean_squared_error")
xgb_grid.fit(x_train, y_train)
print("XGB best:", xgb_grid.best_params_)
XGB_pip = xgb_grid.best_estimator_
evaluate("XGBoost", XGB_pip, x_train, y_train, "Train")
val_scores["XGBoost"] = evaluate("XGBoost", XGB_pip, x_val, y_val, "Val")
```

 # 9️⃣ Store Segmentation (K-Means)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler                         
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# ============ 1) Load ============
# تحميل الداتا النهائية
# Load the final data
df = pd.read_csv("rossmann_fe_final.csv", parse_dates=["date"], dtype={"stateholiday": "string"})
```

```python
# ============ 2) Store-level features ============
# بنحوّل كل فرع لصف واحد فيه "طباعه":
# متوسط المبيعات، التذبذب، حساسية العروض، قوة ديسمبر، هل بيفتح الأحد، بعد المنافس
# We turn each store into one row describing its "behavior":
# average sales, volatility, promo sensitivity, December strength, Sunday opening, competitor distance
g = df.groupby("store")
store_feat = pd.DataFrame({
    "avg_sales": g["sales"].mean(),
    "sales_cv": g["sales"].std() / g["sales"].mean(),   # التذبذب النسبي / relative volatility
})

# حساسية العروض: متوسط المبيعات مع العرض ÷ بدون عرض
# Promo sensitivity: average sales with promo ÷ without promo
store_feat["promo_lift"] = (df[df["promo"] == 1].groupby("store")["sales"].mean()
                            / df[df["promo"] == 0].groupby("store")["sales"].mean())

# قوة ديسمبر: متوسط ديسمبر ÷ متوسط باقي الشهور
# December strength: December average ÷ average of the other months
store_feat["dec_lift"] = (df[df["month"] == 12].groupby("store")["sales"].mean()
                          / df[df["month"] != 12].groupby("store")["sales"].mean())

# هل الفرع بيفتح الأحد (1 أو 0)
# Does the store open on Sundays (1 or 0)
store_feat["sunday_open"] = (df[df["dayofweek"] == 7].groupby("store").size()
                             .reindex(store_feat.index).fillna(0) > 0).astype(int)

# بعد المنافس بـ log علشان القيم الكبيرة ما تسيطرش
# Competitor distance in log so large values do not dominate
store_feat["log_comp_distance"] = np.log1p(g["competitiondistance"].first())

print(store_feat.shape)   # المتوقع / expected: (1115, 6)
```

```python
# ============ 3) Scaling ============
# K-Means بيعتمد على المسافات، فلازم نوحّد المقياس
# K-Means relies on distances, so we must put the features on the same scale
X = StandardScaler().fit_transform(store_feat)
```

```python
# ============ 4) Choose k ============
# Elbow وSilhouette لتجربة k من 2 لـ 8
# Elbow and Silhouette for k from 2 to 8
ks = range(2, 9)
inertia, sil = [], []
for k in ks:
    km = KMeans(n_clusters=k, n_init=10, random_state=42).fit(X)
    inertia.append(km.inertia_)
    sil.append(silhouette_score(X, km.labels_))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(ks, inertia, marker="o"); axes[0].set_title("Elbow (inertia)")
axes[1].plot(ks, sil, marker="o"); axes[1].set_title("Silhouette score")
# 📸 2️⃣ حفظ الصورة باسم جديد ومناسب
plt.savefig("kmeans_elbow_and_silhouette.png", bbox_inches="tight")
plt.tight_layout(); plt.show()
print(dict(zip(ks, np.round(sil, 3))))
```

```python
# ============ 5) Final K-Means ============
# اخترنا k=4 لأن شرائحه مفهومة تجاريًا (اشرح السبب تحت)
# We chose k=4 because its segments make business sense (see the reason below)
km = KMeans(n_clusters=4, n_init=10, random_state=42).fit(X)
store_feat["cluster"] = km.labels_
```

```python
# ============ 6) Name the segments ============
# أرقام الـ clusters عشوائية، فبنسمّيها بالقواعد من الـ Profile مش بالرقم
# Cluster numbers are arbitrary, so we name them by rules from the profile, not by number
prof = store_feat.groupby("cluster").mean()
names = {}
sun = prof["sunday_open"].idxmax();      names[sun] = "Sunday-open stores"
rest = prof.drop(index=sun)
promo = rest["promo_lift"].idxmax();     names[promo] = "Promo-sensitive"
rest = rest.drop(index=promo)
top = rest["avg_sales"].idxmax();        names[top] = "High & stable"
rest = rest.drop(index=top)
names[rest.index[0]] = "Seasonal (Christmas peak)"
store_feat["segment"] = store_feat["cluster"].map(names)
```

```python
# ============ 7) Profile ============
# بروفايل كل شريحة
# Profile of each segment
print(store_feat.groupby("segment").agg(
    stores=("avg_sales", "size"), avg_sales=("avg_sales", "mean"), cv=("sales_cv", "mean"),
    promo_lift=("promo_lift", "mean"), dec_lift=("dec_lift", "mean"),
    sunday=("sunday_open", "mean"), comp=("log_comp_distance", "mean")).round(2))
```

```python
# ============ 8) Save ============
# حفظ الشرائح علشان Power BI
# Save the segments for Power BI
store_feat.reset_index()[["store", "cluster", "segment"]].to_csv("store_segments.csv", index=False)
```

<h3>💡 الفكرة واستراتيجية بناء الميزات (Segmentation Strategy)</h3>
<ul>
  <li>🎯 <b>الهدف:</b> تحويل كافة سجلات المعاملات اليومية لـ <b>1,115 فرعًا</b> إلى ملخص تنفيذي موحد (صف واحد لكل فرع) يضم 6 ميزات تصف سلوك وطبيعة عمل الفرع، ثم تجميع الفروع المتشابهة باستخدام خوارزمية <b>K-Means</b>.</li>

<h2>🏪 Store Segmentation (K-Means Clustering) — Key Insights 💡</h2>

<h3>💡 Clustering Rationale & Feature Architecture</h3>
<ul>
  <li>🎯 <b>Objective:</b> Consolidated daily transaction histories into a store-level behavioral profile (1,115 rows × 6 features) and applied <b>K-Means Unsupervised Learning</b> to segment store archetypes.</li>
  <li>📐 <b>Store-Level Feature Set:</b>
    <ul>
      <li><code>avg_sales</code>: Mean daily gross revenue.</li>

      <li><code>log_comp_distance</code>: Log-transformed distance to nearest competitor.</li>

  </li>

<h3>🔢 Optimal Cluster Selection & Business Utility</h3>
<ul>
  <li>📊 <b>Silhouette Score Analysis:</b> Evaluated cluster ranges from k=2 through k=8. Scores were uniformly low (~0.2), peaking at k=3 (0.235).</li>
  <li>⚠️ <b>Overlapping Boundaries:</b> Low silhouette coefficients indicate soft transitions rather than hard boundaries, treating clusters as <b>dominant behavioral tendencies</b>.</li>
  <li>💡 <b>Strategic Choice of k=4:</b> Selected <b>k=4</b> over k=3 because it effectively decoupled highly seasonal stores from core stable operations, unlocking distinct inventory planning strategies.</li>
  <li>🏷️ <b>Business-Centric Naming:</b> Labeled segments based on domain profile traits instead of volatile cluster ID numbers.</li>

```python
import time
import numpy as np
import pandas as pd
from tensorflow import keras
from tensorflow.keras import layers

# ============ 1) Settings ============
# WINDOW: الموديل بيشوف آخر 28 يوم، GAP: بنتوقع 7 أيام قدام (زي موديل XGBoost بالظبط)
# WINDOW: the model sees the last 28 days, GAP: we predict 7 days ahead (same as the XGBoost model)
WINDOW, GAP, EPOCHS = 28, 7, 8
keras.utils.set_random_seed(42)
```

```python
# ============ 2) Full calendar ============
# بنحتاج الداتا الأصلية بكل الأيام (بما فيها المقفولة) علشان التسلسل الزمني يبقى كامل
# We need the raw data with ALL days (including closed ones) so the time sequence is complete
raw = pd.read_csv("train.csv", low_memory=False, parse_dates=["Date"], dtype={"StateHoliday": "string"})
raw.columns = raw.columns.str.lower()
dates = pd.date_range(raw["date"].min(), raw["date"].max(), freq="D")
n_days, n_stores = len(dates), int(raw["store"].max())
day_idx = (raw["date"] - dates[0]).dt.days.values
store_idx = raw["store"].values - 1

# جدول (يوم × فرع) لكل عمود
# A (day x store) grid for each column
def to_grid(values):
    g = np.zeros((n_days, n_stores), dtype=np.float32)
    g[day_idx, store_idx] = values
    return g

sales = to_grid(raw["sales"].values)
is_open = to_grid((raw["open"].values == 1) & (raw["sales"].values > 0))   # 1 لو الفرع فتح وباع / 1 if the store was open and sold
promo = to_grid(raw["promo"].values)
school = to_grid(raw["schoolholiday"].values)
holiday = to_grid(raw["stateholiday"].values != "0")
```

```python
# ============ 3) Samples ============
# العينات = نفس صفوف الداتا النهائية (النسخة v2)، ونشيل أول 34 يوم لأنهم ملهمش ماضي كفاية
# Samples = the same rows as the final data (v2); we drop the first 34 days because they lack enough history
df = pd.read_csv("rossmann_fe_final_v2.csv", parse_dates=["date"], dtype={"stateholiday": "string"})
df["ti"] = (df["date"] - dates[0]).dt.days.values
df["si"] = df["store"].values - 1
df = df[df["ti"] >= WINDOW + GAP - 1].reset_index(drop=True)

# نفس التقسيم الزمني بالظبط
# Exactly the same time-based split
val_start, test_start = pd.Timestamp("2015-05-08"), pd.Timestamp("2015-06-19")
train_m = df["date"] < val_start
val_m = (df["date"] >= val_start) & (df["date"] < test_start)
test_m = df["date"] >= test_start
print("samples:", train_m.sum(), val_m.sum(), test_m.sum())
```

```python
# ============ 4) Scaling per store ============
# كل فرع له مستوى مبيعات مختلف، فبنطرح متوسط الفرع (من التدريب بس) والموديل يتعلم الانحراف عنه
# Each store has a different sales level, so we subtract the store mean (from train only) and the model learns the deviation
df["log_sales"] = np.log1p(df["sales"])
store_mean = (df[train_m].groupby("si")["log_sales"].mean()
              .reindex(range(n_stores)).fillna(df[train_m]["log_sales"].mean()).values.astype(np.float32))
```

```python
# ============ 5) Sequence cube ============
# لكل يوم وفرع: المبيعات (مطروح منها المتوسط)، هل فتح، العرض، إجازة المدارس، إجازة رسمية، ويوم الأسبوع
# For each day and store: sales (minus the mean), was open, promo, school holiday, state holiday, and weekday
weekday = dates.dayofweek.values
wd_sin = np.broadcast_to(np.sin(2 * np.pi * weekday / 7)[:, None], (n_days, n_stores))
wd_cos = np.broadcast_to(np.cos(2 * np.pi * weekday / 7)[:, None], (n_days, n_stores))
cube = np.stack([(np.log1p(sales) - store_mean[None, :]) * is_open, is_open, promo, school, holiday,
                 wd_sin, wd_cos], axis=-1).astype(np.float32)
offsets = np.arange(-(WINDOW + GAP - 1), -GAP + 1)   # من اليوم -34 لحد اليوم -7 / from day -34 to day -7
```

```python
# ============ 6) Build inputs ============
# 3 مدخلات: التسلسل (آخر 28 يوم)، معلومات اليوم المطلوب (معروفة مقدمًا)، ورقم الفرع
# 3 inputs: the sequence (last 28 days), info about the target day (known in advance), and the store id
def make_inputs(mask):
    d = df[mask]
    ti, si = d["ti"].values, d["si"].values
    x_seq = cube[ti[:, None] + offsets[None, :], si[:, None]]
    month = d["date"].dt.month.values
    x_static = np.column_stack([
        np.eye(7, dtype=np.float32)[d["dayofweek"].values - 1],
        d["promo"], d["schoolholiday"], d["is_state_holiday"], d["promo2_active"],
        d["days_to_next_holiday"] / 14, d["days_since_last_holiday"] / 14,
        d["day_before_holiday"], d["day_after_holiday"],
        np.sin(2 * np.pi * month / 12), np.cos(2 * np.pi * month / 12)]).astype(np.float32)
    y = (d["log_sales"].values - store_mean[si]).astype(np.float32)
    return [x_seq, x_static, si.astype(np.int32)], y, d["sales"].values, store_mean[si]

X_train, y_train, _, _ = make_inputs(train_m)
X_val, y_val, sales_val, mean_val = make_inputs(val_m)
X_test, y_test, sales_test, mean_test = make_inputs(test_m)
print("sequence:", X_train[0].shape, "| static:", X_train[1].shape)   # (734368, 28, 7) و (734368, 17)
```

```python
# ============ 7) Model ============
# LSTM للتسلسل + Dense للمعلومات المعروفة + Embedding للفرع، وبعدين نجمعهم
# LSTM for the sequence + Dense for the known info + Embedding for the store, then we merge them
seq_in = keras.Input((WINDOW, cube.shape[-1]))
static_in = keras.Input((X_train[1].shape[1],))
store_in = keras.Input((), dtype="int32")

h = layers.LSTM(64)(seq_in)
s = layers.Dense(32, activation="relu")(static_in)
e = layers.Flatten()(layers.Embedding(n_stores, 8)(store_in))
z = layers.Concatenate()([h, s, e])
z = layers.Dense(64, activation="relu")(z)
z = layers.Dense(32, activation="relu")(z)
model = keras.Model([seq_in, static_in, store_in], layers.Dense(1)(z))
model.compile(optimizer=keras.optimizers.Adam(1e-3), loss="mse")
print("params:", model.count_params())
```

```python
# ============ 8) Train ============
# EarlyStopping بيوقف التدريب لما الـ Validation يبطل يتحسن، ويرجّع أحسن أوزان
# EarlyStopping stops training when validation stops improving, and restores the best weights
def rmspe(y_true, y_pred):
    return np.sqrt(np.mean(((y_true - y_pred) / y_true) ** 2))

t0 = time.time()
history = model.fit(X_train, y_train, validation_data=(X_val, y_val),
                    epochs=EPOCHS, batch_size=1024, verbose=2,
                    callbacks=[keras.callbacks.EarlyStopping(patience=2, restore_best_weights=True)])
print("training time (sec):", round(time.time() - t0))
```

```python
# ============ 9) Evaluate ============
# بنرجّع المبيعات الحقيقية: expm1(التوقع + متوسط الفرع)
# Convert back to real sales: expm1(prediction + store mean)
for name, X, sales_true, mean in [("Val", X_val, sales_val, mean_val), ("TEST", X_test, sales_test, mean_test)]:
    pred = np.expm1(model.predict(X, batch_size=4096, verbose=0).ravel() + mean)
    print(f"LSTM {name:4s} RMSE={np.sqrt(np.mean((sales_true - pred) ** 2)):7.1f}  "
          f"MAE={np.mean(np.abs(sales_true - pred)):7.1f}  RMSPE={rmspe(sales_true, pred):.4f}")
```

```python
# ============ 10) Save ============
# حفظ الموديل
# Save the model
model.save("rossmann_lstm.keras")
```

<ul>

      <li><b>طبقة Embedding (8 أبعاد):</b> لمعالجة المعرف الخاص بالفرع (Store ID).</li>
      <li><b>التجميع والدمج:</b> تتجمع المخرجات الثلاثة لتمر عبر طبقتي Dense (64 ثم 32) وصولاً لطبقة خروج واحدة (Output Layer).</li>
